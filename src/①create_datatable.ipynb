```python
import pandas as pd
import numpy as np
import glob
```

```python
path = "../data/x4/"
fname = "image_info_220704.csv"
```

```python
file = pd.read_csv(path+fname)
```

```python
file.keys()
```

```python
file
```

```python
cols = ["batchno","day","dish","fname","Mean82","FLAG"]
df = pd.DataFrame([],columns=cols)

count = 0
for index, row in file.iterrows():
    #folder = path + "/Day38/"+ row["撮影日付"] + "/" +row["フォルダ名"] + "/"
    folder = path + "/" +row["フォルダ名"] + "/"
    impathlist = glob.glob(folder+"*.tif")
    #print(impathlist)
    for impath in impathlist:
        array = []

        array.append(row["ラウンド数"])
        array.append(row["培養開始後経過日数(Day xx)"])
        array.append(row["フォルダ名"])
        array.append(impath)
        array.append(row["Day82 Myf5"])
        array.append(row["FLAG"])

        df.loc[count] = array

        count = count +1
```
